```python
import pandas as pd
```

```python
df = pd.read_csv("pilot_results.csv")
print(df.shape)
df.head()
```

```python
results = df[(df["Label"] == "experimental-trial") & (df["Parameter"] == "Selection")]

print(results.shape)
results.head()
```

```python
results.columns
```

```python
results["item_type"] = results["item_id"].apply(lambda x: x[0])
results["expected_response"] = results["comprehension_question"]

results.groupby(["item_type", "expected_response"])["Value"].value_counts()
```

```python
results["orig_flavour"] = results["group"].apply(lambda x: x[0])
results["pos"] = results["group"].apply(lambda x: x.split("_")[1])

results.groupby(["item_type", "expected_response", "orig_flavour"])["Value"].value_counts()
```

```python
results.groupby(["item_type", "expected_response", "pos"])["Value"].value_counts()
```
