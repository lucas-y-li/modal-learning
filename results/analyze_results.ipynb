```python
import pandas as pd
```

```python
df = pd.read_csv("pilot_results.csv")
print(df.shape)
df.head()
```

```python
results = df[(df["Label"] == "experimental-trial") & (df["Parameter"] == "Selection")]
results.head()
```

```python
results.columns
```

```python
results["item_type"] = results["item_id"].apply(lambda x: x[0])
results["expected_response"] = results["comprehension_question"]

results.groupby(["item_type", "expected_response"])["Value"].value_counts()
```
