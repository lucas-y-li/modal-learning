```python
# GENERAL NOTE,,,,,,,,,,,,,,,,,,,,,,,,,,,,DAX is the functional auxiliary nonce form and MIP is the lexical-adjective nonce form in this draft. Counterbalance nonce forms across morphosyntactic conditions in the final experiment. Context components that can appear first in an order variant introduce the participant by name rather than with an ungrounded pronoun; pronouns are retained later where their antecedent is clear.,,,,
# GENERAL NOTE,,,,,,,,,,,,,,,,,,,,,,,,,,,,Context-only and whole-scenario norming must use the same context wording/order for a given item version. Whole-scenario norming differs only by appending the familiar-English target sentence.,,,,
# GENERAL NOTE,,,,,,,,,,,,,,,,,,,,,,,,,,,,"For test/filler items, three order-pair types are balanced exactly: P1=N-E-D + E-D-N; P2=D-N-E + N-D-E; P3=E-N-D + D-E-N. Each pair occurs 8 times, and each of the six individual three-part orders therefore occurs 8 times.",,,,
# GENERAL NOTE,,,,,,,,,,,,,,,,,,,,,,,,,,,,"For training items, each root/deontic candidate has N-D and D-N versions; each epistemic candidate has N-E and E-N versions. The untrained flavor is not made false during training.",,,,
# GENERAL NOTE,,,,,,,,,,,,,,,,,,,,,,,,,,,,"Training candidate syntax is balanced 4/4/4 per flavor across Declarative, Wh-question, and Yes/no inversion. Critical test and filler targets remain declarative to keep the flavor-transfer judgment syntactically conservative. Norming practice/example trials should be declarative.",,,,
# GENERAL NOTE,,,,,,,,,,,,,,,,,,,,,,,,,,,,E_status records whether the target proposition is epistemically possible (E+) or ruled out (E-). epistemic_support_type separately records HOW E+ is established: explicit live alternatives/positive evidence versus merely remaining open because nothing rules it out. Both receive 'Yes' to the context-only epistemic possibility question because both are compatible with the information state.,,,,

df = pd.read_csv("modal_experiment_v1_stimulus_candidates_v8.csv")
df.head()
```

### context-only norming

```python
cols = [
    "phase",
    "item_id",
    "domain",
    "D_status",
    "E_status",
    "temporal_orientation",
    "order_1",
    "order_2",
    "context_order_1",
    "context_order_2",
    "context_only_deontic_question",
    "context_only_epistemic_question",
]
context_only = df[cols]

context_only.head()
```

```python
context_only["context_only_deontic_question"] = context_only["context_only_deontic_question"].apply(
    lambda x: x.replace("rules or permissions", "<b>rules or permissions</b>")
)
context_only["context_only_epistemic_question"] = context_only["context_only_epistemic_question"].apply(
    lambda x: x.replace("information", "<b>information</b>")
)
```

```python
context_only = pd.concat([
        context_only[[
            "phase",
            "item_id",
            "domain",
            "D_status",
            "E_status",
            "temporal_orientation",
            "order_1",
            "context_order_1",
            "context_only_deontic_question",
            "context_only_epistemic_question"]].rename(columns={
                "order_1":"order",
                "context_order_1":"context_order"
                }),
        context_only[[
            "phase",
            "item_id",
            "domain",
            "D_status",
            "E_status",
            "temporal_orientation",
            "order_2",
            "context_order_2",
            "context_only_deontic_question",
            "context_only_epistemic_question"]].rename(columns={
                "order_2":"order",
                "context_order_2":"context_order"
                })
    ])

context_only["D_status"] = context_only["D_status"].map({"D+": 1, "D unspecified": 0, "D-": -1})
context_only["E_status"] = context_only["E_status"].map({"E+ (explicit uncertainty / live alternatives)": 1, "E+ (open / not ruled out)": 0, "E-": -1})

co_d = context_only[[
            "phase",
            "item_id",
            "domain",
            "D_status",
            "temporal_orientation",
            "order",
            "context_order",
            "context_only_deontic_question",
            ]].rename(columns={
                "D_status":"modal_possibility",
                "context_only_deontic_question":"context_only_question"
                })
co_d["judgement_flavour"] = "Deontic"
co_d["context_only_option_3"] = "The story does not say"

co_e = context_only[[
            "phase",
            "item_id",
            "domain",
            "E_status",
            "temporal_orientation",
            "order",
            "context_order",
            "context_only_epistemic_question",
            ]].rename(columns={
                "E_status":"modal_possibility",
                "context_only_epistemic_question":"context_only_question"
                })
co_e["judgement_flavour"] = "Epistemic"
co_e["context_only_option_3"] = ""


context_only = pd.concat([co_d, co_e]).sort_values(by=["item_id", "judgement_flavour", "modal_possibility", "order"]).reset_index(drop=True)

context_only.head()
```

```python
context_only["group"] = (context_only.index + context_only.index // 8) % 8 + 1
context_only["group"].tolist()
```

```python
print(context_only.shape)

context_only.to_csv("context_only_stimulus.csv")
```

### whole-scenario norming

```python
df = pd.read_csv("modal_experiment_v1_stimulus_candidates_v8.csv")
df.head()
```

```python
cols = [
    "phase",
    "item_id",
    "domain",
    "temporal_orientation",
    "order_1",
    "order_2",
    "context_order_1",
    "context_order_2",
    "whole_scenario_may_target", "whole_scenario_possible_target",
    "whole_scenario_deontic_reading", "whole_scenario_epistemic_reading",
    "expected_whole_scenario_deontic_interpretation", "expected_whole_scenario_epistemic_interpretation",
    "syntax"
]

whole_scenario = df[cols]

whole_scenario["whole_scenario_question"] = whole_scenario["whole_scenario_may_target"].apply(
    lambda x: "Which interpretation better matches what the speaker is talking about?" if x[-1] == "." \
        else "Which interpretation better matches what the speaker is asking about?"
)

whole_scenario.head()
```

```python
whole_scenario = pd.concat([
        whole_scenario[[
            "phase",
            "item_id",
            "domain",
            "temporal_orientation",
            "order_1",
            "context_order_1",
            "whole_scenario_may_target", "whole_scenario_possible_target",
            "whole_scenario_question",
            "whole_scenario_deontic_reading", "whole_scenario_epistemic_reading",
            "expected_whole_scenario_deontic_interpretation", "expected_whole_scenario_epistemic_interpretation",
            "syntax"
            ]].rename(columns={
                "order_1":"order",
                "context_order_1":"context_order"
                }),
        whole_scenario[[
            "phase",
            "item_id",
            "domain",
            "temporal_orientation",
            "order_2",
            "context_order_2",
            "whole_scenario_may_target", "whole_scenario_possible_target",
            "whole_scenario_question",
            "whole_scenario_deontic_reading", "whole_scenario_epistemic_reading",
            "expected_whole_scenario_deontic_interpretation", "expected_whole_scenario_epistemic_interpretation",
            "syntax"
            ]].rename(columns={
                "order_2":"order",
                "context_order_2":"context_order"
                }),
    ])


whole_scenario.head()
```

```python
whole_scenario_may = whole_scenario[[
        "phase",
        "item_id",
        "domain",
        "temporal_orientation",
        "order",
        "context_order",
        "whole_scenario_may_target",
        "whole_scenario_question",
        "whole_scenario_deontic_reading", "whole_scenario_epistemic_reading",
        "expected_whole_scenario_deontic_interpretation", "expected_whole_scenario_epistemic_interpretation",
        "syntax"
        ]].rename(columns={
            "whole_scenario_may_target":"whole_scenario_response",
        })
whole_scenario_may["response_type"] = "may"

whole_scenario_possible = whole_scenario[[
        "phase",
        "item_id",
        "domain",
        "temporal_orientation",
        "order",
        "context_order",
        "whole_scenario_possible_target",
        "whole_scenario_question",
        "whole_scenario_deontic_reading", "whole_scenario_epistemic_reading",
        "expected_whole_scenario_deontic_interpretation", "expected_whole_scenario_epistemic_interpretation",
        "syntax"
        ]].rename(columns={
            "whole_scenario_possible_target":"whole_scenario_response",
        })
whole_scenario_may["response_type"] = "possible"


whole_scenario_may.head()
```

```python
# participants see either may or possible, not both. 

whole_scenario_may["group"] = (whole_scenario_may.index + whole_scenario_may.index // 2) % 2 + 1
print(whole_scenario_may["group"].tolist())

whole_scenario_possible["group"] = (whole_scenario_possible.index + whole_scenario_possible.index // 2) % 2 + 3
print(whole_scenario_possible["group"].tolist())
```

```python
whole_scenario = pd.concat([whole_scenario_may, whole_scenario_possible])

whole_scenario["whole_scenario_response"] = whole_scenario["whole_scenario_response"].apply(
    lambda x: x.split(" — ")[0])

print(whole_scenario.shape)
whole_scenario.head()

```

```python
# make it so that which interpretation is on the left/right depends on the group of the person.

whole_scenario_de = whole_scenario.copy()
whole_scenario_de["left_interpretation"] = whole_scenario_de["whole_scenario_deontic_reading"]
whole_scenario_de["right_interpretation"] = whole_scenario_de["whole_scenario_epistemic_reading"]
whole_scenario_de["scale_order"] = "deontic-epistemic"

whole_scenario_ed = whole_scenario.copy()
whole_scenario_ed["left_interpretation"] = whole_scenario_ed["whole_scenario_epistemic_reading"]
whole_scenario_ed["right_interpretation"] = whole_scenario_ed["whole_scenario_deontic_reading"]
whole_scenario_ed["scale_order"] = "epistemic-deontic"
whole_scenario_ed["group"] = whole_scenario_ed["group"] + 4
print(whole_scenario_ed["group"].tolist())

whole_scenario_combined = pd.concat([whole_scenario_de, whole_scenario_ed])

whole_scenario_combined["left_interpretation"] = whole_scenario_combined["left_interpretation"].apply(
    lambda x: x.replace(
        "rules or permissions", "<span style='color:#C62828'>rules or permissions</span>"
        ).replace(
            "available information", "<span style='color:#C62828'>available information</span>"
        )
)

whole_scenario_combined["right_interpretation"] = whole_scenario_combined["right_interpretation"].apply(
    lambda x: x.replace(
        "rules or permissions", "<span style='color:#1565C0'>rules or permissions</span>"
        ).replace(
            "available information", "<span style='color:#1565C0'>available information</span>"
        )
)

whole_scenario_combined.to_csv("whole_scenario_stimulus.csv")

```

```python
# also update the practice trials to use the same orders

whole_scenario_practice = pd.read_csv("norming_practice_trials_whole_scenario.csv")

practice_groups = []
for group in range(1, 9):
    p = whole_scenario_practice.copy()
    p["group"] = group
    if group > 4:
        p["left_interpretation"] = p["whole_scenario_epistemic_reading"]
        p["right_interpretation"] = p["whole_scenario_deontic_reading"]
        p["expected_preference"] = p["expected_preference"].apply(
            lambda x: '5' if x == '1' else '1' if x == '5' else x
        )
        p["scale_order"] = "epistemic-deontic"
    else:
        p["left_interpretation"] = p["whole_scenario_deontic_reading"]
        p["right_interpretation"] = p["whole_scenario_epistemic_reading"]
        p["scale_order"] = "deontic-epistemic"
    practice_groups.append(p)

whole_scenario_practice_grouped = pd.concat(practice_groups)

whole_scenario_practice_grouped["left_interpretation"] = whole_scenario_practice_grouped["left_interpretation"].apply(
    lambda x: x.replace(
        "rules or permissions", "<span style='color:#C62828'>rules or permissions</span>"
        ).replace(
            "available information", "<span style='color:#C62828'>available information</span>"
        )
)

whole_scenario_practice_grouped["right_interpretation"] = whole_scenario_practice_grouped["right_interpretation"].apply(
    lambda x: x.replace(
        "rules or permissions", "<span style='color:#1565C0'>rules or permissions</span>"
        ).replace(
            "available information", "<span style='color:#1565C0'>available information</span>"
        )
)

whole_scenario_practice_grouped
```
