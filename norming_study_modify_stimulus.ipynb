### context-only norming

```python
cols = [
    "phase",
    "item_id",
    "domain",
    "D_status",
    "E_status",
    "temporal_orientation",
    "order_1",
    "order_2",
    "context_order_1",
    "context_order_2",
    "context_only_deontic_question",
    "context_only_epistemic_question",
]
context_only = df[cols]

context_only.head()
```

```python
context_only["context_only_deontic_question"] = context_only["context_only_deontic_question"].apply(
    lambda x: x.replace("rules or permissions", "<b>rules or permissions</b>")
)
context_only["context_only_epistemic_question"] = context_only["context_only_epistemic_question"].apply(
    lambda x: x.replace("information", "<b>information</b>")
)
```

```python
context_only = pd.concat([
        context_only[[
            "phase",
            "item_id",
            "domain",
            "D_status",
            "E_status",
            "temporal_orientation",
            "order_1",
            "context_order_1",
            "context_only_deontic_question",
            "context_only_epistemic_question"]].rename(columns={
                "order_1":"order",
                "context_order_1":"context_order"
                }),
        context_only[[
            "phase",
            "item_id",
            "domain",
            "D_status",
            "E_status",
            "temporal_orientation",
            "order_2",
            "context_order_2",
            "context_only_deontic_question",
            "context_only_epistemic_question"]].rename(columns={
                "order_2":"order",
                "context_order_2":"context_order"
                })
    ])

context_only["D_status"] = context_only["D_status"].map({"D+": 1, "D unspecified": 0, "D-": -1})
context_only["E_status"] = context_only["E_status"].map({"E+ (explicit uncertainty / live alternatives)": 1, "E+ (open / not ruled out)": 0, "E-": -1})

co_d = context_only[[
            "phase",
            "item_id",
            "domain",
            "D_status",
            "temporal_orientation",
            "order",
            "context_order",
            "context_only_deontic_question",
            ]].rename(columns={
                "D_status":"modal_possibility",
                "context_only_deontic_question":"context_only_question"
                })
co_d["judgement_flavour"] = "Deontic"
co_d["context_only_option_3"] = "The story does not say"

co_e = context_only[[
            "phase",
            "item_id",
            "domain",
            "E_status",
            "temporal_orientation",
            "order",
            "context_order",
            "context_only_epistemic_question",
            ]].rename(columns={
                "E_status":"modal_possibility",
                "context_only_epistemic_question":"context_only_question"
                })
co_e["judgement_flavour"] = "Epistemic"
co_e["context_only_option_3"] = ""


context_only = pd.concat([co_d, co_e]).sort_values(by=["item_id", "judgement_flavour", "modal_possibility", "order"]).reset_index(drop=True)

context_only.head()
```

```python
context_only["group"] = (context_only.index + context_only.index // 8) % 8 + 1
context_only["group"].tolist()
```

```python
print(context_only.shape)

context_only.to_csv("context_only_stimulus.csv")
```

### whole-scenario norming

```python
df = pd.read_csv("modal_experiment_v1_stimulus_candidates_v8.csv")
# The distributed CSV includes GENERAL NOTE rows without an item_id.
df = df.loc[df["item_id"].notna()].copy()
df.head()
```

```python
cols = [
    "phase",
    "item_id",
    "domain",
    "temporal_orientation",
    "order_1",
    "order_2",
    "context_order_1",
    "context_order_2",
    "whole_scenario_may_target", "whole_scenario_possible_target",
    "whole_scenario_deontic_reading", "whole_scenario_epistemic_reading",
    "expected_whole_scenario_deontic_interpretation", "expected_whole_scenario_epistemic_interpretation",
    "syntax"
]

whole_scenario = df[cols].copy()

whole_scenario["whole_scenario_question"] = whole_scenario["whole_scenario_may_target"].apply(
    lambda x: "What is the speaker talking about?" if x[-1] == "." \
        else "What is the speaker asking about?"
)

whole_scenario.head()
```

```python
# Preserve each actual context and its order label; code the two order versions.
context_versions = []
for context_code, suffix in enumerate([1, 2]):
    version = whole_scenario.drop(columns=[
        f"order_{3 - suffix}", f"context_order_{3 - suffix}"
    ]).rename(columns={
        f"order_{suffix}": "order",
        f"context_order_{suffix}": "context_order",
    }).copy()
    version["context_order_code"] = context_code
    context_versions.append(version)

whole_scenario = pd.concat(context_versions, ignore_index=True)
whole_scenario.head()
```

```python
whole_scenario_may = whole_scenario[[
        "phase",
        "item_id",
        "domain",
        "temporal_orientation",
        "order",
        "context_order",
        "context_order_code",
        "whole_scenario_may_target",
        "whole_scenario_question",
        "whole_scenario_deontic_reading", "whole_scenario_epistemic_reading",
        "expected_whole_scenario_deontic_interpretation", "expected_whole_scenario_epistemic_interpretation",
        "syntax"
        ]].rename(columns={
            "whole_scenario_may_target":"whole_scenario_response",
        })
whole_scenario_may["response_type"] = "may"

whole_scenario_possible = whole_scenario[[
        "phase",
        "item_id",
        "domain",
        "temporal_orientation",
        "order",
        "context_order",
        "context_order_code",
        "whole_scenario_possible_target",
        "whole_scenario_question",
        "whole_scenario_deontic_reading", "whole_scenario_epistemic_reading",
        "expected_whole_scenario_deontic_interpretation", "expected_whole_scenario_epistemic_interpretation",
        "syntax"
        ]].rename(columns={
            "whole_scenario_possible_target":"whole_scenario_response",
        })
whole_scenario_possible["response_type"] = "possible"


whole_scenario_may.head()
```

```python
whole_scenario = pd.concat([whole_scenario_may, whole_scenario_possible]).reset_index(drop=True)

whole_scenario["whole_scenario_response"] = whole_scenario["whole_scenario_response"].apply(
    lambda x: x.split(" — ")[0])

print(whole_scenario.shape)
whole_scenario.head()

```

```python
# 2 response types x 2 scale orders x 4 rotating item lists = 16 groups.
# Response and scale order are fixed within a group. The temporal/context
# combination rotates across items, with one actual version per item per group.
import re

variants = whole_scenario.rename(columns={"context_order": "context"}).copy()
variants["context_order"] = variants["context_order_code"]
temporal_codes = {
    "future eventive": 0,
    "past perfect / present progressive or stative": 1,
    "past perfect": 1,
    "present progressive": 1,
    "present stative": 1,
}
variants["temporal_code"] = variants["temporal_orientation"].map(temporal_codes)
assert variants["temporal_code"].notna().all(), "Unrecognized temporal orientation"
variants["temporal_code"] = variants["temporal_code"].astype(int)

# Natural ordering gives C1, C2, ... C8, F1, ... F8, T1, ... T12.
def item_sort_key(item_id):
    return tuple(int(part) if part.isdigit() else part
                 for part in re.split(r"(\d+)", str(item_id)))

item_ids = sorted(variants["item_id"].unique(), key=item_sort_key)
item_rank = {item_id: rank for rank, item_id in enumerate(item_ids)}
variants["item_rank"] = variants["item_id"].map(item_rank)
variant_keys = ["item_id", "response_type", "temporal_code", "context_order"]
assert not variants.duplicated(variant_keys).any(), "Duplicate stimulus version"
available_temporal = variants.groupby("item_id")["temporal_code"].agg(lambda x: set(x))
paired_items = [item for item in item_ids if available_temporal[item] == {0, 1}]
paired_rank = {item: rank for rank, item in enumerate(paired_items)}
# Some fillers have only one temporal version in the source CSV. Keep that
# version, and balance context order within each fixed temporal category.
fixed_items = [item for item in item_ids if len(available_temporal[item]) == 1]
fixed_context_rank = {}
for temporal_code in [0, 1]:
    items = [item for item in fixed_items if available_temporal[item] == {temporal_code}]
    fixed_context_rank.update({item: rank for rank, item in enumerate(items)})

expanded_groups = []
group_definitions = []
for response_code, response_type in enumerate(["may", "possible"]):
    for scale_code, scale_order in enumerate(["deontic-epistemic", "epistemic-deontic"]):
        for list_rotation in range(4):
            group = response_code * 8 + scale_code * 4 + list_rotation + 1
            assignments = []
            for item in item_ids:
                if item in paired_rank:
                    combination = (paired_rank[item] + list_rotation) % 4
                    temporal_code, context_order = divmod(combination, 2)
                else:
                    temporal_code = next(iter(available_temporal[item]))
                    context_order = (fixed_context_rank[item] + list_rotation) % 2
                assignments.append({
                    "item_id": item, "response_type": response_type,
                    "temporal_code": temporal_code, "context_order": context_order,
                })
            selected = pd.DataFrame(assignments).merge(
                variants, on=variant_keys, how="left", validate="one_to_one", indicator=True
            )
            assert selected["_merge"].eq("both").all(), "Missing actual stimulus version"
            selected = selected.drop(columns="_merge")
            selected["group"] = group
            selected["response_code"] = response_code
            selected["scale_code"] = scale_code
            selected["scale_order"] = scale_order
            selected["list_rotation"] = list_rotation
            # Use the item-specific reading endpoints from the selected source row.
            left = "whole_scenario_deontic_reading" if scale_code == 0 else "whole_scenario_epistemic_reading"
            right = "whole_scenario_epistemic_reading" if scale_code == 0 else "whole_scenario_deontic_reading"
            selected["left_interpretation"] = selected[left]
            selected["right_interpretation"] = selected[right]
            expanded_groups.append(selected)
            group_definitions.append({
                "group": group, "response_type": response_type,
                "scale_order": scale_order, "scale_code": scale_code,
                "list_rotation": list_rotation,
            })

group_conditions = pd.DataFrame(group_definitions)
whole_scenario_combined = pd.concat(expanded_groups, ignore_index=True).sort_values(
    ["group", "item_rank"], kind="mergesort"
).drop(columns="item_rank").reset_index(drop=True)
```

```python
# Check assignment and counterbalancing before exporting.
assert set(whole_scenario_combined["group"]) == set(range(1, 17))
item_group_counts = whole_scenario_combined.groupby(["item_id", "group"]).size()
assert item_group_counts.eq(1).all(), "An item repeats within a group"
assert whole_scenario_combined.groupby("item_id")["group"].nunique().eq(16).all()
assert whole_scenario_combined.groupby("group")["response_type"].nunique().eq(1).all()
assert whole_scenario_combined.groupby("group")["scale_order"].nunique().eq(1).all()

combination_counts = whole_scenario_combined.groupby(
    ["group", "temporal_code", "context_order"]
).size().unstack(["temporal_code", "context_order"], fill_value=0)
assert set(combination_counts.columns) == {(0, 0), (0, 1), (1, 0), (1, 1)}
assert combination_counts.gt(0).all().all(), "A group is missing a combination"
# Fixed-temporal fillers can make the temporal totals unequal. Within each
# temporal category, context orders must still be as balanced as possible.
for temporal_code in [0, 1]:
    counts = combination_counts[temporal_code]
    assert (counts.max(axis=1) - counts.min(axis=1)).le(1).all(), "Unbalanced context orders"

# Each paired item sees all four combinations across the four lists of each
# response/scale pair. Fixed-temporal fillers see each context order twice.
paired_counts = whole_scenario_combined.loc[
    whole_scenario_combined["item_id"].isin(paired_items)
].groupby(["item_id", "response_type", "scale_order", "temporal_code", "context_order"]).size()
assert len(paired_counts) == len(paired_items) * 16 and paired_counts.eq(1).all()
fixed_counts = whole_scenario_combined.loc[
    whole_scenario_combined["item_id"].isin(fixed_items)
].groupby(["item_id", "response_type", "scale_order", "context_order"]).size()
assert len(fixed_counts) == len(fixed_items) * 8 and fixed_counts.eq(2).all()

whole_scenario_combined.to_csv("whole_scenario_stimulus.csv", index=False)
print("Verified 16 groups, one version per item/group, and balanced combinations.")
combination_counts
```

```python
# Practice trials use the same scale order as their experimental group.
whole_scenario_practice = pd.read_csv("norming_practice_trials_whole_scenario.csv")
practice_groups = []
for condition in group_conditions.itertuples(index=False):
    p = whole_scenario_practice.copy()
    p["group"] = condition.group
    p["scale_order"] = condition.scale_order
    p["response_type"] = condition.response_type
    left = "whole_scenario_deontic_reading" if condition.scale_code == 0 else "whole_scenario_epistemic_reading"
    right = "whole_scenario_epistemic_reading" if condition.scale_code == 0 else "whole_scenario_deontic_reading"
    p["left_interpretation"] = p[left]
    p["right_interpretation"] = p[right]
    if condition.scale_code == 1:
        # Labels naming the interpretation stay unchanged. Reverse numeric
        # left/right answer codes only (CSV may load them as strings or numbers).
        p["expected_preference"] = p["expected_preference"].replace({
            "1": "2", "2": "1", 1: 2, 2: 1,
        })
    practice_groups.append(p)

whole_scenario_practice_grouped = pd.concat(practice_groups, ignore_index=True)
practice_scale = whole_scenario_practice_grouped.groupby("group")["scale_order"].first()
assert practice_scale.equals(group_conditions.set_index("group")["scale_order"])
whole_scenario_practice_grouped
```
