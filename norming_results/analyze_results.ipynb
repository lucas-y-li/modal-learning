```python
df = pd.read_csv("whole_scenario_results.csv")
print(df.shape)
df.head()
```

```python
results = df[(df["Label"] == "experimental-trial") & (df["Parameter"] == "Selection")]
results["selected"] = results.apply(
    lambda x: x["left_interpretation"] if x["Value"] == "left" else x["right_interpretation"],
    axis=1
)

results.head()
```

```python
import scipy.stats

def binom_test(results, groups, value):
    arr = results.groupby([groups])[value].value_counts().reset_index().to_numpy().reshape(-1, 2, 3)
    d = []
    for i in range(arr.shape[0]):
        group_name = arr[i][0][0]
        assert group_name == arr[i][1][0]

        if arr[i][0][1] == "The available information":
            epistemic_count = arr[i][0][2]
            deontic_count = arr[i][1][2]
        else:
            epistemic_count = arr[i][1][2]
            deontic_count = arr[i][0][2]
    
        result = scipy.stats.binomtest(k=epistemic_count, n=epistemic_count + deontic_count, p=0.5, alternative='two-sided')
        d.append({
            groups: group_name, 
            "count": epistemic_count+deontic_count, 
            "epistemic_percent": epistemic_count/(epistemic_count+deontic_count),
            "p-value": result.pvalue,
            "significant": "True" if result.pvalue < 0.05 else "",
        })
    df = pd.DataFrame(d)
    return df

binom_test(results, "order", "selected")
```

```python
from statsmodels.genmod.bayes_mixed_glm import BinomialBayesMixedGLM

df = results.copy()

# 1 = epistemic response, 0 = deontic response
df["epistemic_response"] = (
    df["selected"] == "The available information"
).astype(int)

# Your current coding:
# future eventive = expected deontic
# everything else = expected epistemic
df["expected_flavor"] = df["temporal_orientation"].apply(
    lambda x: "deontic" if x == "future eventive" else "epistemic"
)

# Make reference levels explicit:
# deontic target = reference
# may = reference
df["expected_flavor"] = pd.Categorical(
    df["expected_flavor"],
    categories=["deontic", "epistemic"]
)

df["response_condition"] = pd.Categorical(
    df["response_condition"],
    categories=["may", "possible"]
)
df["participant_id"] = df["MD5.hash.of.participant.s.IP.address"]


model = BinomialBayesMixedGLM.from_formula(
    "epistemic_response ~ C(expected_flavor) * C(response_condition)",
    {
        "participant": "0 + C(participant_id)",
        "item": "0 + C(item_id)"
    },
    df
)

fit = model.fit_vb()

print(fit.summary())
```

```python
df_no_wh = df[df["response_syntax"] != "Wh-question"].copy()

model_no_wh = BinomialBayesMixedGLM.from_formula(
    "epistemic_response ~ C(expected_flavor) * C(response_condition)",
    {
        "participant": "0 + C(participant_id)",
        "item": "0 + C(item_id)"
    },
    df_no_wh
)

fit_no_wh = model_no_wh.fit_vb()

print(fit_no_wh.summary())
```

```python
summary = (
    df_no_wh
    .groupby(
        ["expected_flavor", "response_condition"],
        observed=True
    )
    .agg(
        n=("epistemic_response", "size"),
        epistemic_percent=("epistemic_response", "mean")
    )
    .reset_index()
)

print(summary)
```

```python
pivot = summary.pivot(
    index="response_condition",
    columns="expected_flavor",
    values="epistemic_percent"
)

pivot["flavor_discrimination"] = (
    pivot["epistemic"] - pivot["deontic"]
)

print(pivot)
```
