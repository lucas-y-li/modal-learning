```python
import pandas as pd
```

```python
df = pd.read_csv("context_only_results_force.csv")
print(df.shape)
df.head()
```

```python
results = df[(df["Label"] == "experimental-trial") & (df["Parameter"] == "Selection")]
results.head()
```

```python
results["phase"] = results["item_id"].apply(lambda x: x[0])
results["phase"]
```

```python
results["group"].value_counts()
```

```python
results.groupby(["judgment_force", "modal_polarity"])["Value"].value_counts()
```
