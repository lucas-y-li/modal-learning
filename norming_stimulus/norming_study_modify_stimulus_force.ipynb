```python
import pandas as pd
```

```python
df = pd.read_csv("../modal_experiment_v1_negative_force_stimulus_candidates.csv")
df = df.loc[df["item_id"].notna()].copy()
df.head()
```

### context-only norming

```python
cols = [
    "phase",
    "item_id",
    "domain",
    "P_status",
    "Nec_status",
    "temporal_orientation",
    "order_1",
    "order_2",
    "context_order_1",
    "context_order_2",
    "context_only_possibility_question",
    "context_only_necessity_question",
    "direction_or_condition"
]
context_only = df[cols]

context_only.head()
```

```python
context_only = pd.concat([
        context_only[[
            "phase",
            "item_id",
            "domain",
            "P_status",
            "Nec_status",
            "temporal_orientation",
            "order_1",
            "context_order_1",
            "context_only_possibility_question",
            "context_only_necessity_question", 
            "direction_or_condition"]].rename(columns={
                "order_1":"order",
                "context_order_1":"context_order"
                }),
        context_only[[
            "phase",
            "item_id",
            "domain",
            "P_status",
            "Nec_status",
            "temporal_orientation",
            "order_2",
            "context_order_2",
            "context_only_possibility_question",
            "context_only_necessity_question",
            "direction_or_condition"]].rename(columns={
                "order_2":"order",
                "context_order_2":"context_order"
                })
    ])

context_only["P_status"] = context_only["P_status"].map({"P+": 1, "P-": -1})
context_only["Nec_status"] = context_only["Nec_status"].map({"Nec+": 1, "Nec-": -1})
context_only["flavour"] = context_only["direction_or_condition"].apply(
    lambda x: "D" if "Root" in x else "E"
)

co_p = context_only[[
            "phase",
            "item_id",
            "domain",
            "P_status",
            "temporal_orientation",
            "order",
            "context_order",
            "context_only_possibility_question",
            ]].rename(columns={
                "P_status":"modal_polarity",
                "context_only_possibility_question":"context_only_question"
                })
co_p["judgement_force"] = "possibility"

co_n = context_only[[
            "phase",
            "item_id",
            "domain",
            "Nec_status",
            "temporal_orientation",
            "order",
            "context_order",
            "context_only_necessity_question",
            ]].rename(columns={
                "Nec_status":"modal_polarity",
                "context_only_necessity_question":"context_only_question"
                })
co_n["judgement_force"] = "necessity"

context_only = pd.concat([co_p, co_n]).sort_values(by=["item_id", "judgement_force", "modal_polarity", "order"]).reset_index(drop=True)
context_only.head(10)
```

```python
context_only["group"] = (context_only.index + context_only.index // 8) % 8 + 1
context_only["group"].tolist()
```

```python
print(context_only.shape)

context_only.to_csv("context_only_force_stimulus.csv")
```
